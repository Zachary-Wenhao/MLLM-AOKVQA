```python
import matplotlib.pyplot as plt
import numpy as np

# Data
models = ['GPT-4o', '+ cross attention', '+ chain of thoughts', '+ few shots learning']
multiple_choice = [54.59, 53.36, 53.54, 52.4]
direct_answer = [36.16, 36.68, 37.21, 36.68]
plt.figure(figsize=(10,7))

x = np.arange(len(models))  # the label locations
width = 0.35  # the width of the bars

fig, ax1 = plt.subplots(figsize=(10,9))

# Bar plot for Multiple Choice (left axis)
color = (0,0,0.3)
# ax1.set_xlabel('Models')
ax1.set_ylabel('Multiple Choice (%)', color=color)
bars = ax1.bar(x - width/2, multiple_choice, width, label='Multiple Choice', color="tab:blue", alpha = 1)
ax1.tick_params(axis='y', labelcolor="tab:blue")
ax1.set_ylim(50, 56)  # Set range for Multiple Choice

# Line plot for Multiple Choice trend
ax1.plot(x - width/2, multiple_choice, color='darkblue', marker='o', linestyle='--', linewidth=2)

# Create second y-axis for Direct Answer
ax2 = ax1.twinx()
color = (0.3,0,0)
ax2.set_ylabel('Direct Answer (%)', color=color)
ax2.plot(x + width/2, direct_answer, color='darkred', marker='s', linestyle='--', linewidth=2)
ax2.tick_params(axis='y', labelcolor=color)
ax2.set_ylim(35, 38)  # Set range for Direct Answer

# Bar plot for Direct Answer (right axis)
bars2 = ax2.bar(x + width/2, direct_answer, width, label='Direct Answer', color=(0.7,0.25,0.3), alpha=1)

# Add some text for labels, title and custom x-axis tick labels
ax1.set_title('Model Accuracy', pad=20, fontsize = 25)
ax1.set_xticks(x)
ax1.set_xticklabels(models, fontsize = 15)

# Combine legends from both axes
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc='upper right', fontsize = 25)

# Add value labels on bars
for bar in bars:
    height = bar.get_height()
    ax1.annotate(f'{height:.2f}',
                 xy=(bar.get_x() + bar.get_width() / 2, height),
                 xytext=(0, 3),  # 3 points vertical offset
                 textcoords="offset points",
                 ha='center', va='bottom')

for bar in bars2:
    height = bar.get_height()
    ax2.annotate(f'{height:.2f}',
                 xy=(bar.get_x() + bar.get_width() / 2, height),
                 xytext=(0, 3),  # 3 points vertical offset
                 textcoords="offset points",
                 ha='center', va='bottom')

fig.tight_layout()
plt.show()
```
